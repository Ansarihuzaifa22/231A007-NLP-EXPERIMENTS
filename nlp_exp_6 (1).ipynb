```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents

nltk.download('treebank')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets_json')
```

```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
text = "On his off duty period, enjoying vacation, he met Martha and the two started a new life together. At one point during their life together, he had to leave for another war that was declared on the country. Martha waited patiently for his return, and Arthur did return. Soon, they got married and moved in together, owning a house in Dimen City."
tokens = word_tokenize(text)
print(tokens)
```

```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
rule_based_tag = pos_tag(tokens)
print(rule_based_tag)
```

```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
train_sents = treebank.tagged_sents()[:3000]

unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)

bigram_tagger = BigramTagger(train_sents, backoff = unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)

print("Unigram Tags: ", unigram_tags)
print("Bigram Tags: ", bigram_tags)
```

```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
print("Original text: ")
print(text)
print("\nRule-based Tagging (Statistical):")
print(rule_based_tag)
print("\nDoug was here...")
print("Unigram Tagging:")
print(unigram_tags)
print("\nBigram Tagging:")
print(bigram_tags)
```

```python
student = "Huzaifa ansari | UIN: 231A007"
print(student)
test_sents = treebank.tagged_sents()[3000:]
print(f"Accuracy of Unigram Tagger: {unigram_tagger.accuracy(test_sents) * 100:.2f}%")
print(f"Accuracy of Bigram Tagger: {bigram_tagger.accuracy(test_sents) * 100:.2f}%")
```
